**POSTFIX CODE - state**

```python
def postfix_pda(s):
    state = "q0"
    stack = ["Z"]

    state = "q1"

    for x in s.replace(" ", ""):

        if x == "n":
            stack.append("N")

        elif x == "o" and stack.count("N") >= 2:
            stack.pop()
            stack.pop()
            stack.append("N")

        else:
            return False

    if state == "q1" and stack == ["Z", "N"]:
        state = "q2"

    return state == "q2"

s = input("Masukkan postfix: ")
print("DITERIMA" if postfix_pda(s) else "DITOLAK")
```

**PREFIX CODE - state**

```python
def prefix_pda(s):
    state = "q0"
    stack = ["Z"]

    stack.append("A")
    state = "q1"

    for x in s.replace(" ", ""):
        if x == "o" and stack[-1] in ["A", "X"]:
            stack.append("X")

        elif x == "n" and stack[-1] in ["A", "X"]:
            stack.pop()

        else:
            return False

    if state == "q1" and stack == ["Z"]:
        state = "q2"

    return state == "q2"

s = input("Masukkan prefix: ")
print("DITERIMA" if prefix_pda(s) else "DITOLAK")
```

**POSTFIX & PREFIX - state**

```python
def cek(s):
    s = s.replace(" ", "")

    def valid(teks):
        butuh = 1
        for x in teks:
            if butuh == 0 or x not in "on":
                return False
            butuh += 1 if x == "o" else -1
        return butuh == 0

    pre = valid(s)
    post = valid(s[::-1])

    if pre and post:
        return "DITERIMA sebagai PREFIX dan POSTFIX"
    if pre:
        return "DITERIMA sebagai PREFIX"
    if post:
        return "DITERIMA sebagai POSTFIX"
    return "DITOLAK"

print(cek(input("Masukkan ekspresi: ")))
```
